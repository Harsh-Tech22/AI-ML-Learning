```
import pandas as pd

data = {
    "Name": ["Harsh", "Aniket", "Devika", "Banti", "Mayank"],
    "Age": [19, 18, 18, 19, 30],
    "Marks": [85, 78, 92, 88, 75],
    "Salary": [25000, 30000, 35000, 28000, 32000]
}

pd.DataFrame(data).to_csv("data.csv", index=False)

df = pd.read_csv("data.csv")

print("Complete Dataset:")
print(df)

print("\nNumber of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])

print("\nFirst Five Rows:")
print(df.head())

print("\nSize of Dataset:", df.size)

print("\nMissing Values:")
print(df.isnull().sum())

numeric_df = df.select_dtypes(include="number")

print("\nSum of Numerical Columns:")
print(numeric_df.sum())

print("\nAverage of Numerical Columns:")
print(numeric_df.mean())

print("\nMinimum Values:")
print(numeric_df.min())

print("\nMaximum Values:")
print(numeric_df.max())

df.to_csv("exported_data.csv", index=False)

print("\nData exported successfully!")
```
